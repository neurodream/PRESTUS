```python
def free_field_intensity_cm2(pressure_pa, density_kgm3=1000, speed_of_sound_ms=1500):
    """
    Calculate the free-field intensity of a single ultrasound wave in W/cm².

    Parameters:
    pressure_pa (float): Pressure amplitude at the transducer exit in Pascals.
    density_kgm3 (float, optional): Density of the medium (default: 1000 kg/m³ for water).
    speed_of_sound_ms (float, optional): Speed of sound in the medium (default: 1500 m/s for water).

    Returns:
    float: Intensity in W/cm².
    """
    intensity_wm2 = (pressure_pa ** 2) / (density_kgm3 * speed_of_sound_ms)
    intensity_wcm2 = intensity_wm2 / 10_000  # Convert W/m² to W/cm²
    return intensity_wcm2
```

```python
# Example usage
pressure_pa = 231440 # 1 MPa = 1,000,000 Pa
intensity_wcm2 = free_field_intensity_cm2(pressure_pa)
print(f"Intensity: {intensity_wcm2:.2f} W/cm²")
```
