```python
import numpy as np
import pandas as pd
import os
import matplotlib.pyplot as plt
from pathlib import Path
import matplotlib.cm as cm
```

```python
path = Path("../data/transducer_profiles")
files = [path/f for f in os.listdir(path) if f.startswith("Axial_profiles_IS")]
files
```

```python
for f in files:

    df = pd.read_csv(f, skiprows=1)

    cols = df.columns
    colors = cm.viridis(np.linspace(0, 1, len(cols[1:])))
    plt.figure(figsize=(12,3))
    for i, col in enumerate(cols[1:]):
        plt.plot(df[cols[0]], df[col], color=colors[i])
    plt.title(f.stem.replace("Axial_profiles_IS_", "").replace("~IGT_128_ch_comb_10_ch", ""))
    plt.show()
```

```python
# sham
for f in files:
    df = pd.read_csv(f, skiprows=1)
    print(f.stem.replace("Axial_profiles_IS_", "sham_").replace("~IGT_128_ch_comb_10_ch", ""), "=", df[df.columns[1]].tolist(), ";")
# active
for f in files:
    df = pd.read_csv(f, skiprows=1)
    print(f.stem.replace("Axial_profiles_IS_", "active_").replace("~IGT_128_ch_comb_10_ch", ""), "=", df[df.columns[-1]].tolist(), ";")
```

```python
plt.figure(figsize=(12,3))
labels = []
for f in files:

    df = pd.read_csv(f, skiprows=1)
    
    plt.plot(df["Distance WRT Exit plane (mm)"], df["90"])
    labels.append(f.stem.replace("Axial_profiles_IS_", "").replace("~IGT_128_ch_comb_10_ch", ""))
plt.legend(labels)
plt.show()
```
